```python
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import matplotlib.colors as mcolors
```

```python
# Function to create a colormap from a AR6 text file containing RGB values
def create_colormap_from_txt(filepath, cmap_name="custom_colormap"):
    """
    Read RGB values from a text file and create a LinearSegmentedColormap.
    RGB value shoud be 0-1 range.
    """
    rgb_data = np.loadtxt(filepath)
    colormap = mcolors.LinearSegmentedColormap.from_list(cmap_name, rgb_data)
    return colormap
```

```python
# Function to plot a colorbar
def plot_colorbar(cmap, title):
    fig, ax = plt.subplots(figsize=(8, 1.2))
    fig.subplots_adjust(bottom=0.5)

    norm = mpl.colors.Normalize(vmin=0, vmax=100)
    cb = mpl.colorbar.ColorbarBase(
        ax,
        cmap=cmap,
        norm=norm,
        orientation='horizontal'
    )

    # Remove ticks and labels
    cb.set_ticks([])
    
    # Remove outline (box)
    cb.outline.set_visible(False)

    # Optional: also remove axis spines completely
    ax.set_frame_on(False)

    ax.set_title(title, fontsize=12)
    plt.show()
```

```python
# Temperature (Divergent) 
temp_div_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/temp_div.txt")

# Temperature (Sequential)
temp_seq_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/temp_seq.txt")

#plot
# plot_colorbar(temp_div_colormap, "Temperature Diverging Colormap from File")
# plot_colorbar(temp_seq_colormap, "Temperature Sequential Colormap from File")

plot_colorbar(temp_div_colormap,' ')
plot_colorbar(temp_seq_colormap, ' ')
```
