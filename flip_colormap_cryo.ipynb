```python
import numpy as np

# load the colormap
data = np.loadtxt("./continuous_colormaps_rgb_0-1/cryo_div.txt")

# reverse the order of rows
data_flipped = data[::-1]

# save back to a new file
np.savetxt("./continuous_colormaps_rgb_0-1/cryo_div_flipped.txt", data_flipped, fmt="%.6f")
```

```python
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import matplotlib.colors as mcolors

# Function to create a colormap from a AR6 text file containing RGB values
def create_colormap_from_txt(filepath, cmap_name="custom_colormap"):
    """
    Read RGB values from a text file and create a LinearSegmentedColormap.
    RGB value shoud be 0-1 range.
    """
    rgb_data = np.loadtxt(filepath)
    colormap = mcolors.LinearSegmentedColormap.from_list(cmap_name, rgb_data)
    return colormap

# Function to plot a colorbar
def plot_colorbar(cmap, title):
    fig, ax = plt.subplots(figsize=(8, 1.2))
    fig.subplots_adjust(bottom=0.5)

    norm = mpl.colors.Normalize(vmin=0, vmax=100)
    cb = mpl.colorbar.ColorbarBase(
        ax,
        cmap=cmap,
        norm=norm,
        orientation='horizontal'
    )

    ax.set_title(title, fontsize=12)
    plt.show()
```

```python
# --- Cryosphere (Divergent) ---
cryo_div_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/cryo_div.txt")

cryo_div_flipped_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/cryo_div_flipped.txt")

#plot
plot_colorbar(cryo_div_colormap, "Cryosphere Diverging Colormap from File")
plot_colorbar(cryo_div_flipped_colormap, "Cryosphere Diverging Colormap Flipped from File")
```
