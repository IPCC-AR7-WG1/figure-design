```python
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import matplotlib.colors as mcolors
```

```python
# Function to create a colormap from a AR6 text file containing RGB values
def create_colormap_from_txt(filepath, cmap_name="custom_colormap"):
    """
    Read RGB values from a text file and create a LinearSegmentedColormap.
    RGB value shoud be 0-1 range.
    """
    rgb_data = np.loadtxt(filepath)
    colormap = mcolors.LinearSegmentedColormap.from_list(cmap_name, rgb_data)
    return colormap
```

```python
import os
import re

def plot_colorbar(
    cmap,
    filename=None,
    folder="./colormap_bar",
    vmin=0,
    vmax=100,
    orientation="horizontal",
    transparent=True,
    show=False
):
    # ---- Resolve filename ----
    if filename is None:
        filename = getattr(cmap, "name", "colormap")

    # Sanitize filename
    filename = re.sub(r"[^a-zA-Z0-9_-]", "_", filename)

    # Ensure folder exists
    os.makedirs(folder, exist_ok=True)

    save_path = os.path.join(folder, f"{filename}.svg")

    # ---- Create figure ----
    fig, ax = plt.subplots(figsize=(8, 1.2 if orientation == "horizontal" else 4))

    try:
        norm = mpl.colors.Normalize(vmin=vmin, vmax=vmax)

        cb = mpl.colorbar.ColorbarBase(
            ax,
            cmap=cmap,
            norm=norm,
            orientation=orientation
        )

        # Clean look
        cb.set_ticks([])
        cb.outline.set_visible(False)
        ax.set_frame_on(False)

        fig.tight_layout()

        # ---- Save ----
        fig.savefig(
            save_path,
            format="svg",
            bbox_inches="tight",
            transparent=transparent
        )

        print(f"Saved to: {save_path}")

        if show:
            plt.show()

    finally:
        plt.close(fig)
```

```python
# Temperature (Divergent) 
temp_div_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/temp_div.txt")

# Temperature (Sequential)
temp_seq_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/temp_seq.txt")

#plot
# plot_colorbar(temp_div_colormap, "Temperature Diverging Colormap from File")
# plot_colorbar(temp_seq_colormap, "Temperature Sequential Colormap from File")

plot_colorbar(temp_div_colormap,'temp_div')
plot_colorbar(temp_seq_colormap, 'temp_seq')
```
