```python
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import matplotlib.colors as mcolors
```

```python
# Function to create a colormap from a AR6 text file containing RGB values
def create_colormap_from_txt(filepath, cmap_name="custom_colormap"):
    """
    Read RGB values from a text file and create a LinearSegmentedColormap.
    RGB value shoud be 0-1 range.
    """
    rgb_data = np.loadtxt(filepath)
    colormap = mcolors.LinearSegmentedColormap.from_list(cmap_name, rgb_data)
    return colormap
```

```python
# Function to plot a colorbar
def plot_colorbar(cmap, title):
    fig, ax = plt.subplots(figsize=(8, 1.2))
    fig.subplots_adjust(bottom=0.5)

    norm = mpl.colors.Normalize(vmin=0, vmax=100)
    cb = mpl.colorbar.ColorbarBase(
        ax,
        cmap=cmap,
        norm=norm,
        orientation='horizontal'
    )

    ax.set_title(title, fontsize=12)
    plt.show()
```

```python
# --- Temperature (Divergent: blue → white → red) --- Ref: colorbrewer 
temp_div = LinearSegmentedColormap.from_list(
    "temp_div",
    ["#053061",'#2166ac', "#4393c3",'#92c5de', "#d1e5f0",'#f7f7f7', '#fddbc7', "#f4a582", "#d6604d", "#b2182b", '#67001f']
)

temp_div_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/temp_div.txt")

# --- Temperature (Sequential: yellow → orange → dark brown) ---
temp_seq = 'afmhot_r' # reverse of 'afmhot'

temp_seq_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/temp_seq.txt")

#plot
plot_colorbar(temp_div, "Temperature (Divergent)")

plot_colorbar(temp_div_colormap, "Temperature Diverging Colormap from File")

plot_colorbar(temp_seq, "Temperature (Sequential)")

plot_colorbar(temp_seq_colormap, "Temperature Sequential Colormap from File")
```

```python
# --- Precipitation (Divergent: brown → white → teal) ---
prec_div = 'BrBG'

prec_div_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/prec_div.txt")

# --- Precipitation (Sequential: light green → dark teal) ---
brbg = plt.get_cmap('BrBG')
green_1 = brbg(0.75)
green_2 = brbg(0.95) 
green_3 = brbg(1.0)

prec_seq = LinearSegmentedColormap.from_list(
    "prec_seq",
    ["#e5f5e0", "#99d8c9", green_1, green_2, green_3]
)

prec_seq_colormap = create_colormap_from_txt("./continuous_colormaps_rgb_0-1/prec_seq.txt")

#plot
plot_colorbar(prec_div, "Precipitation (Divergent)")
plot_colorbar(prec_div_colormap, "Precipitation Diverging Colormap from File")
plot_colorbar(prec_seq, "Precipitation (Sequential)")
plot_colorbar(prec_seq_colormap, "Precipitation Sequential Colormap from File")
```
