# convert RGB .svg to CMYK .pdf

Example of using the *svg_to_cmyk.py* module to convert SVG files into CMYK PDFs.

```python
from svg_to_cmyk import SVGToCMYKConverter

converter = SVGToCMYKConverter()
```

```python
# Convert a single SVG file to CMYK PDF
converter.svg_to_cmyk_pdf("./fig_svg/fig_1map.svg", "fig_cmyk_pdf/fig_1map_cmyk.pdf")
```

```python
# Batch convert all SVG files in a folder to CMYK PDF
converter.batch_convert("fig_svg", "fig_cmyk_pdf")
```
